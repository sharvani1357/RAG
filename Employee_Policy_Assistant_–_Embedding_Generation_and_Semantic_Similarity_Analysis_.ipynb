Semantic Similarity Analysis

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

chunks = [
    "Employees receive 12 casual leaves annually.",
    "Employees receive 15 sick leaves annually.",
    "Employees may work from home twice per week.",
    "Travel expenses are reimbursed within 30 days.",
    "All employees are covered under company medical insurance."
]

queries = [
    "How many casual leaves are allowed?",
    "Can employees work remotely?",
    "What is the travel reimbursement process?",
    "Do employees have medical insurance?"
]

model = SentenceTransformer("all-MiniLM-L6-v2")

chunk_embeddings = model.encode(chunks)

query_embeddings = model.encode(queries)

print("-"*50)
print("SEMANTIC SIMILARITY ANALYSIS")
print("-"*50)

for i, query in enumerate(queries):
    print(f"\nQuery: {query}\n")

    similarities = cosine_similarity(
        [query_embeddings[i]],
        chunk_embeddings
    )[0]

    for j, chunk in enumerate(chunks):
        print(f"Chunk: {chunk}")
        print(f"Similarity Score: {similarities[j]:.4f}\n")

print("-"*50)
```

Identify Most Similar Chunk

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

chunks = [
    "Employees receive 12 casual leaves annually.",
    "Employees receive 15 sick leaves annually.",
    "Employees may work from home twice per week.",
    "Travel expenses are reimbursed within 30 days.",
    "All employees are covered under company medical insurance."
]

queries = [
    "How many casual leaves are allowed?",
    "Can employees work remotely?",
    "What is the travel reimbursement process?",
    "Do employees have medical insurance?"
]

model = SentenceTransformer("all-MiniLM-L6-v2")

chunk_embeddings = model.encode(chunks)

query_embeddings = model.encode(queries)

print("-"*50)
print("MOST SIMILAR CHUNK RETRIEVAL")
print("-"*50)

for i, query in enumerate(queries):
    similarities = cosine_similarity(
        [query_embeddings[i]],
        chunk_embeddings
    )[0]

    best_index = np.argmax(similarities)

    print(f"\nQuery: {query}")
    print(f"Most Similar Chunk: {chunks[best_index]}")
    print(f"Similarity Score: {similarities[best_index]:.4f}")

print("-"*50)
```

Similar Meaning vs Different Meaning

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

sentence1 = "Employees receive 12 casual leaves."
sentence2 = "Workers are entitled to 12 annual leaves."

sentence3 = "Employees receive 12 casual leaves."
sentence4 = "Travel expenses are reimbursed within 30 days."

model = SentenceTransformer("all-MiniLM-L6-v2")

embedding1 = model.encode(sentence1)
embedding2 = model.encode(sentence2)

embedding3 = model.encode(sentence3)
embedding4 = model.encode(sentence4)

similarity1 = cosine_similarity([embedding1], [embedding2])[0][0]
similarity2 = cosine_similarity([embedding3], [embedding4])[0][0]

print("-"*50)
print("SEMANTIC SIMILARITY COMPARISON")
print("-"*50)

print("\nSentence Pair 1")
print(sentence1)
print(sentence2)
print("Cosine Similarity:")
print(round(similarity1, 4))

print("\nSentence Pair 2")
print(sentence3)
print(sentence4)
print("Cosine Similarity:")
print(round(similarity2, 4))

print("-"*50)
```

Embedding Visualization using PCA

```python
from sentence_transformers import SentenceTransformer
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

chunks = [
    "Employees receive 12 casual leaves annually.",
    "Employees receive 15 sick leaves annually.",
    "Employees may work from home twice per week.",
    "Travel expenses are reimbursed within 30 days.",
    "All employees are covered under company medical insurance."
]

queries = [
    "How many casual leaves are allowed?",
    "Can employees work remotely?",
    "What is the travel reimbursement process?",
    "Do employees have medical insurance?"
]

model = SentenceTransformer("all-MiniLM-L6-v2")

chunk_embeddings = model.encode(chunks)

query_embeddings = model.encode(queries)

all_embeddings = list(chunk_embeddings) + list(query_embeddings)

pca = PCA(n_components=2)

reduced_embeddings = pca.fit_transform(all_embeddings)

plt.figure(figsize=(10, 6))

for i in range(len(chunks)):
    plt.scatter(
        reduced_embeddings[i, 0],
        reduced_embeddings[i, 1]
    )
    plt.annotate(
        f"C{i+1}",
        (reduced_embeddings[i, 0], reduced_embeddings[i, 1])
    )

for i in range(len(queries)):
    idx = i + len(chunks)

    plt.scatter(
        reduced_embeddings[idx, 0],
        reduced_embeddings[idx, 1])

    plt.annotate(
        f"Q{i+1}",
        (reduced_embeddings[idx, 0], reduced_embeddings[idx, 1])
    )

plt.title("PCA Visualization of Chunk and Query Embeddings")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)

plt.show()
```
