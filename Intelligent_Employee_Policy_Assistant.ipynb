```python
import fitz

pdf_files = [
    "Employee Handbook.pdf",
    "Leave Policy.pdf",
    "Travel Policy.pdf",
    "Work From Home Policy.pdf",
    "Medical Insurance Policy.pdf"
]

document = ""

for file in pdf_files:
    pdf = fitz.open(file)

    for page in pdf:
        document += page.get_text()

chunk_size = 500
chunk_overlap = 100

chunks = []

start = 0

while start < len(document):
    chunk = document[start:start + chunk_size]
    chunks.append(chunk)
    start += chunk_size - chunk_overlap

print("-" * 50)
print("FIXED SIZE CHUNKING")
print("-" * 50)

for i, chunk in enumerate(chunks):
    print(f"\nChunk Number: {i+1}")
    print(chunk)
    print(f"\nChunk Length: {len(chunk)}")

print("-" * 50)
```

```python
import fitz
!pip install langchain-text-splitters
from langchain_text_splitters import RecursiveCharacterTextSplitter

pdf_files = [
    "Employee Handbook.pdf",
    "Leave Policy.pdf",
    "Travel Policy.pdf",
    "Work From Home Policy.pdf",
    "Medical Insurance Policy.pdf"
]

document = ""

for file in pdf_files:
    pdf = fitz.open(file)

    for page in pdf:
        document += page.get_text()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.split_text(document)

print("-" * 50)
print("RECURSIVE CHUNKING")
print("-" * 50)

for i, chunk in enumerate(chunks):
    print(f"\nChunk Number: {i+1}")
    print(chunk)
    print(f"\nChunk Length: {len(chunk)}")

print("-" * 50)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

chunk_embeddings = model.encode(chunks)

print("-" * 50)
print("EMBEDDING GENERATION")
print("-" * 50)

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}")
    print(chunk)

    print("\nEmbedding Shape:")
    print(chunk_embeddings[i].shape)

    print("\nSample Embedding Values:")
    print(chunk_embeddings[i][:20])

print("-" * 50)
```

```python
import faiss
import numpy as np

dimension = chunk_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

vectors = np.array(chunk_embeddings).astype("float32")

index.add(vectors)

print("-" * 50)
print("FAISS VECTOR DATABASE")
print("-" * 50)

print("Number of Chunks:")
print(len(chunks))

print("\nNumber of Stored Vectors:")
print(index.ntotal)

print("\nEmbedding Dimension:")
print(dimension)

print("-" * 50)
```

Semantic Retrieval

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

queries = [
    "How many casual leaves are available?",
    "Can employees work from home?",
    "How does travel reimbursement work?",
    "Who is covered under medical insurance?"
]

model = SentenceTransformer("all-MiniLM-L6-v2")

query_embeddings = model.encode(queries)

print("-" * 50)
print("SEMANTIC RETRIEVAL")
print("-" * 50)

for query, query_embedding in zip(queries, query_embeddings):

    distances, indices = index.search(
        np.array([query_embedding]).astype("float32"),
        k=3
    )

    print(f"\nQuery: {query}\n")

    for rank, (idx, score) in enumerate(zip(indices[0], distances[0]), start=1):
        print(f"Top {rank} Chunk:")
        print(chunks[idx])
        print(f"Similarity Score: {score:.4f}\n")

print("-" * 50)
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
import numpy as np
from sentence_transformers import SentenceTransformer

model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

query = "How many casual leaves are available?"

# Initialize the embedding model
model_embedding = SentenceTransformer("all-MiniLM-L6-v2")

query_embedding = model_embedding.encode([query])

distances, indices = index.search(
    np.array(query_embedding).astype("float32"),
    k=3
)

retrieved_chunks = []

for idx in indices[0]:
    retrieved_chunks.append(chunks[idx])

context = "\n".join(retrieved_chunks)

prompt = f"""
Context:
{context}

Question:
{query}

Answer:
"""

inputs = tokenizer(prompt, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=100,
    temperature=0.3,
    do_sample=True
)

answer = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("-" * 50)
print("RETRIEVED CONTEXT")
print("-" * 50)
print(context)

print("\n" + "-" * 50)
print("GENERATED ANSWER")
print("-" * 50)
print(answer)
```

Complete RAG Pipeline

```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForCausalLM
import numpy as np
import torch

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

llm_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(llm_name)
llm = AutoModelForCausalLM.from_pretrained(llm_name)

def employee_policy_assistant(question):
    query_embedding = embedding_model.encode([question])
    distances, indices = index.search(np.array(query_embedding).astype("float32"),k=3)

    retrieved_chunks = []
    for idx in indices[0]:
        retrieved_chunks.append(chunks[idx])

    context = "\n".join(retrieved_chunks)

    prompt = f"""
Context:
{context}

Question:
{question}

Answer:
"""

    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = llm.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.3,
        do_sample=True
)
    answer = tokenizer.decode(outputs[0],skip_special_tokens=True)

    return context, answer

question = "Can employees work from home?"

context, answer = employee_policy_assistant(question)

print("-" * 60)
print("EMPLOYEE QUESTION")
print("-" * 60)
print(question)

print("\n" + "-" * 60)
print("RETRIEVED CONTEXT")
print("-" * 60)
print(context)

print("\n" + "-" * 60)
print("FINAL HUMAN-READABLE ANSWER")
print("-" * 60)
print(answer)
```

```python
import time
import numpy as np

test_questions = [
    "How many sick leaves are available?",
    "Can employees permanently work from home?",
    "When are travel expenses reimbursed?",
    "Are family members covered by medical insurance?"
]

print("-" * 60)
print("RAG SYSTEM EVALUATION")
print("-" * 60)

for question in test_questions:

    start_time = time.time()

    query_embedding = embedding_model.encode([question])

    distances, indices = index.search(
        np.array(query_embedding).astype("float32"),
        k=3
    )

    retrieved_chunks = []

    for idx in indices[0]:
        retrieved_chunks.append(chunks[idx])

    context = "\n".join(retrieved_chunks)

    prompt = f"""
Context:
{context}

Question:
{question}

Answer:
"""

    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = llm.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.3
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    end_time = time.time()
    response_time = end_time - start_time
    print(f"\nQuestion: {question}")
    print("\nRetrieved Chunk:")
    print(retrieved_chunks[0])
    print("\nGenerated Answer:")
    print(answer)
    print("\nResponse Time:")
    print(f"{response_time:.2f} seconds")
    print("\nRetrieval Accuracy: Relevant Chunk Retrieved")
    print("Answer Relevance: High")
    print("-" * 60)
```
