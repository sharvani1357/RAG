```python
text = """
Employees receive 12 casual leaves annually.
Employees can work from home twice per week.
Travel expenses are reimbursed within 30 days.
Medical insurance is provided to all employees.
"""
```

###Fixed-size chunking

```python
chunk_size=50
chunks = [text[i:i+chunk_size] for i in range(0, len(text), chunk_size)]
for chunk in chunks:
  print(chunk)
  print('-'*50)
```

i.e., texts is divided into chunks each having 50 chars

###Recursive chunking

```python
!pip install -q langchain langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10
)
chunks=splitter.split_text(text)
for i,chunk in enumerate(chunks):
  print(f"chunk {i+1}")
  print(chunk)
  print('-'*50)
```

###Basic sentence chunking using NLTK

```python
import nltk
nltk.download('punkt_tab') # Download 'punkt_tab' as indicated by the error
from nltk.tokenize import sent_tokenize

# Split into Sentences
chunks = sent_tokenize(text)

for i, chunk in enumerate(chunks):
    print(f'Chunk {i+1}')
    print(chunk)
    print('-' * 50)
```

```python
from nltk.tokenize import sent_tokenize

# Split into Sentences
chunks = sent_tokenize(text)

for i, chunk in enumerate(chunks):
    print(f'Chunk {i+1}:{chunk}')
```

```python
sentences = sent_tokenize(text)
chunk_size = 2
chunks = []

for i in range(0, len(sentences), chunk_size):
    chunk = ' '.join(sentences[i:i+chunk_size])
    chunks.append(chunk)

for i, chunk in enumerate(chunks, start=1):
    print(f'Chunk {i}:{chunk}')
```

Semantic Chunking Using Sentence Tranformers

```python
!pip install sentence.transformers
```

```python
sentences=sent_tokenize(text)
chunk_size=2
chunks=[]
for i in range(0,len(sentences),chunk_size):
  chunk=' '.join(sentences[i:i+chunk_size])
  chunks.append(chunk)
for i,chunk in enumerate(chunks,start=1):
  print(f"\nchunk {i}:{chunk}")
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
sentences=sent_tokenize(text)
model=SentenceTransformer('all-MiniLM-L6-v2')
embeddings=model.encode(sentences)
threshold=0.5
chunks=[]
current_chunk=[sentences[0]]
for i in range(1,len(sentences)):
  similarity=cosine_similarity([embeddings[i-1]],[embeddings[i]])[0][0]
  if similarity<threshold:
    chunks.append(' '.join(current_chunk))
  else:
    chunks.append(' '.join(current_chunk))
    current_chunk=[sentences[i]]
chunks.append(' '.join(current_chunk))
for id,chunk in enumerate(chunks,start=1):
  print(f'\nChunk {id}:')
  print(chunk)
```
